## Линейная модель

Трудоёмкость задачи в часах: $f_1$ — число файлов, которые придётся изменить,
$f_2 = 1$, если нужен новый функционал. При $\boldsymbol{\theta} = (2;\ 0{,}5;\ 3)$ задача
на 10 файлов с новым функционалом оценивается так:

```python
theta = np.array([2, 0.5, 3])
x = np.array([1, 10, 1])      # f_0 = 1 — фиктивный признак для свободного члена
print(theta @ x)              # скалярное произведение ⟨θ, x⟩
```

## Функция потерь и эмпирический риск

Две модели на четырёх объектах с выбросом (табл. 1.3): A: $g(x) = x$ и Б: $g(x) = 1{,}5x$.
С абсолютной потерей лучше модель A, с квадратичной — модель Б.

```python
x = np.array([1, 2, 3, 4])
y = np.array([1, 2, 3, 8])
for name, pred in [("A", x), ("Б", 1.5 * x)]:
    err = pred - y
    print(name, "абсолютная:", np.mean(np.abs(err)), " квадратичная:", np.mean(err ** 2))
```

```python
t = np.linspace(0, 5, 2)
fig, ax = plt.subplots(figsize=(5, 3.5))
ax.plot(t, t, label="модель A: $g(x) = x$")
ax.plot(t, 1.5 * t, "--", label="модель Б: $g(x) = 1{,}5x$")
ax.scatter(x, y, color="black", zorder=3, label="объекты выборки")
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.grid(True)
ax.legend()
plt.show()
```

Разница между потерей и функционалом качества: потеря — число на каждом объекте,
эмпирический риск — одно число на всю выборку.

```python
err = np.array([1, -2, 0])
print("потери:", err ** 2, " эмпирический риск:", np.mean(err ** 2))
```

## Типы задач

Классификация — граница между классами, регрессия — кривая, проходящая близко к точкам
(рис. 1.2, синтетические данные).

```python
rng = np.random.default_rng(7)
c1 = rng.normal([1.0, 1.0], 0.55, size=(25, 2))
c2 = rng.normal([2.6, 2.4], 0.55, size=(25, 2))
xr = np.sort(rng.uniform(0, 10, 40))
yr = 0.08 * xr ** 2 + 0.3 * xr + rng.normal(0, 0.8, xr.size)

fig, (a, b) = plt.subplots(1, 2, figsize=(9, 3.5))
a.scatter(*c1.T, marker="o", facecolors="white", edgecolors="C0", label="класс −1")
a.scatter(*c2.T, marker="x", color="C1", label="класс +1")
a.plot([-0.5, 4], [4, -0.5], color="black")
a.set_title("а) классификация")
a.set_xlabel("$f_1(x)$")
a.set_ylabel("$f_2(x)$")
a.legend()
t = np.linspace(0, 10, 200)
b.scatter(xr, yr, facecolors="white", edgecolors="C0")
b.plot(t, 0.08 * t ** 2 + 0.3 * t, color="C1")
b.set_title("б) регрессия")
b.set_xlabel("$f_1(x)$")
b.set_ylabel("$y$")
plt.show()
```

## Задачи

**1.** Эмпирический риск с индикатором ошибки для классификатора
$g(x) = \operatorname{sign}(f_2(x) - 10)$ на выборке табл. 1.1.

```python
g = np.sign(data["сложность"] - 10)
print("ошибки на:", list(data.index[g != data["y"]]))
print("Q =", np.mean(g != data["y"]))
```

**2.** Параметр модели $g(x) = \theta x$, минимизирующий квадратичный эмпирический риск
на выборке табл. 1.3: $\theta = \sum x_i y_i / \sum x_i^2$.

```python
theta = np.sum(x * y) / np.sum(x ** 2)
print(np.sum(x * y), "/", np.sum(x ** 2), "=", round(theta, 3))
```
